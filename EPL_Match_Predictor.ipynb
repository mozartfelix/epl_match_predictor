```python

# ==========================================
# TAHAP 2: DOWNLOAD DATASET EPL
# Sumber: Football-Data.co.uk
# Periode: Musim 2015/2016 - 2024/2025
# ==========================================

import pandas as pd
import requests
from io import BytesIO

# Daftar kode 10 musim EPL
seasons = [
    "1516", "1617", "1718", "1819", "1920",
    "2021", "2122", "2223", "2324", "2425"
]

# Tempat menampung data dari semua musim
all_matches = []

# Mengunduh satu per satu dataset setiap musim
for season in seasons:

    # Membentuk URL dataset berdasarkan kode musim
    url = f"https://www.football-data.co.uk/mmz4281/{season}/E0.csv"

    print(f"Mengunduh musim {season}...")

    try:
        response = requests.get(url, timeout=30)
        response.raise_for_status()

        # Membaca file CSV
        df = pd.read_csv(
            BytesIO(response.content),
            encoding="latin1"
        )

        # Merapikan nama kolom
        df.columns = df.columns.str.strip()

        # Memastikan kolom penting tersedia
        required_columns = [
            "Date", "HomeTeam", "AwayTeam",
            "FTHG", "FTAG", "FTR"
        ]

        missing_columns = [
            col for col in required_columns
            if col not in df.columns
        ]

        if missing_columns:
            print(
                f"Kolom tidak tersedia pada musim {season}: "
                f"{missing_columns}"
            )
            continue

        # Menambahkan informasi musim
        df["Season"] = season

        # Menyimpan data musim ini
        all_matches.append(df)

        print(f"Berhasil: {len(df)} baris data.")

    except Exception as e:
        print(f"Gagal mengunduh musim {season}: {e}")

# Menggabungkan seluruh musim
if len(all_matches) == 0:
    raise ValueError(
        "Tidak ada dataset berhasil diunduh. "
        "Periksa koneksi internet dan URL."
    )

epl_data = pd.concat(all_matches, ignore_index=True)

# Mengurutkan data berdasarkan tanggal pertandingan
epl_data["Date"] = pd.to_datetime(
    epl_data["Date"],
    dayfirst=True,
    errors="coerce"
)

epl_data = epl_data.sort_values(
    "Date"
).reset_index(drop=True)

# Menyimpan dataset gabungan ke file CSV
file_name = "epl_matches_2015_2025.csv"
epl_data.to_csv(file_name, index=False)

print("\n==========================================")
print("DOWNLOAD DATASET SELESAI")
print("==========================================")
print(f"Jumlah pertandingan: {len(epl_data)}")
print(f"Jumlah kolom: {len(epl_data.columns)}")
print(f"File tersimpan: {file_name}")
```

```python

# TAHAP 3.4 — MEMERIKSA VALIDITAS KOLOM

kolom_wajib = [
    "Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG", "FTR"
]

kolom_tidak_ada = [
    kolom for kolom in kolom_wajib
    if kolom not in epl_data.columns
]

if kolom_tidak_ada:
    raise ValueError(
        f"Kolom wajib tidak ditemukan: {kolom_tidak_ada}"
    )

print("Semua kolom wajib tersedia.")

# Pastikan kolom skor berbentuk angka
epl_data["FTHG"] = pd.to_numeric(
    epl_data["FTHG"], errors="coerce"
)
epl_data["FTAG"] = pd.to_numeric(
    epl_data["FTAG"], errors="coerce"
)

# Periksa nilai kosong pada kolom utama
print("\nNilai kosong pada kolom utama:")
display(epl_data[kolom_wajib].isnull().sum().to_frame("Jumlah Kosong"))

# Periksa skor yang tidak valid: kosong, negatif, atau bukan bilangan bulat
skor_invalid = (
    epl_data["FTHG"].isna()
    | epl_data["FTAG"].isna()
    | (epl_data["FTHG"] < 0)
    | (epl_data["FTAG"] < 0)
    | (epl_data["FTHG"] % 1 != 0)
    | (epl_data["FTAG"] % 1 != 0)
)

print("\nJumlah baris dengan skor tidak valid:", skor_invalid.sum())

# Periksa kode hasil pertandingan
kode_hasil_valid = ["H", "D", "A"]

hasil_invalid = (
    epl_data["FTR"].isna()
    | ~epl_data["FTR"].isin(kode_hasil_valid)
)

print("Jumlah baris dengan kode FTR tidak valid:", hasil_invalid.sum())

# Hitung hasil yang tidak cocok dengan skor
hasil_dari_skor = pd.Series(
    "D", index=epl_data.index, dtype="object"
)

hasil_dari_skor.loc[
    epl_data["FTHG"] > epl_data["FTAG"]
] = "H"

hasil_dari_skor.loc[
    epl_data["FTHG"] < epl_data["FTAG"]
] = "A"

hasil_tidak_cocok = (
    epl_data["FTHG"].notna()
    & epl_data["FTAG"].notna()
    & epl_data["FTR"].isin(kode_hasil_valid)
    & (epl_data["FTR"] != hasil_dari_skor)
)

print(
    "Jumlah hasil FTR yang tidak cocok dengan skor:",
    hasil_tidak_cocok.sum()
)

# Periksa nama tim kosong atau hanya berisi spasi
tim_invalid = (
    epl_data["HomeTeam"].isna()
    | epl_data["AwayTeam"].isna()
    | epl_data["HomeTeam"].astype("string").str.strip().eq("")
    | epl_data["AwayTeam"].astype("string").str.strip().eq("")
)

print("Jumlah baris dengan nama tim kosong:", tim_invalid.sum())

# Tampilkan beberapa baris yang perlu diperiksa jika ditemukan
baris_perlu_diperiksa = (
    skor_invalid
    | hasil_invalid
    | hasil_tidak_cocok
    | tim_invalid
    | epl_data["Date"].isna()
)

print(
    "\nTotal baris yang perlu diperiksa:",
    baris_perlu_diperiksa.sum()
)

if baris_perlu_diperiksa.any():
    display(
        epl_data.loc[
            baris_perlu_diperiksa,
            kolom_wajib
        ].head(20)
    )
else:
    print("Semua pemeriksaan validitas utama berhasil.")
```

```python

# TAHAP 4.2 — MEMBUAT FITUR PERFORMA 5 PERTANDINGAN SEBELUMNYA

# Setiap tim memiliki riwayat statistik pertandingan
riwayat_tim = defaultdict(list)

data_fitur = []

# Proses pertandingan per tanggal, dari paling lama ke paling baru
for tanggal, pertandingan_harian in epl_clean.groupby("Date", sort=True):

    # Tahap A: buat fitur menggunakan riwayat SEBELUM tanggal ini
    fitur_harian = []

    for idx, laga in pertandingan_harian.iterrows():
        kandang = laga["HomeTeam"]
        tandang = laga["AwayTeam"]

        # Ambil lima pertandingan terakhir masing-masing tim
        riwayat_kandang = riwayat_tim[kandang][-5:]
        riwayat_tandang = riwayat_tim[tandang][-5:]

        # Jika salah satu tim belum punya 5 laga sebelumnya,
        # pertandingan ini belum digunakan untuk training.
        if len(riwayat_kandang) < 5 or len(riwayat_tandang) < 5:
            continue

        fitur = {
            "Date": tanggal,
            "HomeTeam": kandang,
            "AwayTeam": tandang,

            # Fitur performa tim kandang
            "HomeGoalsAvg5": sum(
                laga_lalu["goals_for"] for laga_lalu in riwayat_kandang
            ) / 5,

            "HomeConcededAvg5": sum(
                laga_lalu["goals_against"] for laga_lalu in riwayat_kandang
            ) / 5,

            "HomePointsAvg5": sum(
                laga_lalu["points"] for laga_lalu in riwayat_kandang
            ) / 5,

            # Fitur performa tim tandang
            "AwayGoalsAvg5": sum(
                laga_lalu["goals_for"] for laga_lalu in riwayat_tandang
            ) / 5,

            "AwayConcededAvg5": sum(
                laga_lalu["goals_against"] for laga_lalu in riwayat_tandang
            ) / 5,

            "AwayPointsAvg5": sum(
                laga_lalu["points"] for laga_lalu in riwayat_tandang
            ) / 5,

            # Target: hasil aktual pertandingan ini
            "FTR": laga["FTR"]
        }

        fitur_harian.append(fitur)

    # Simpan fitur tanggal ini
    data_fitur.extend(fitur_harian)

    # Tahap B: setelah fitur hari ini dibuat,
    # baru masukkan hasil pertandingan hari ini ke riwayat.
    # Dengan demikian, tidak ada hasil pertandingan hari ini
    # yang digunakan untuk memprediksi pertandingan pada hari yang sama.
    for _, laga in pertandingan_harian.iterrows():
        kandang = laga["HomeTeam"]
        tandang = laga["AwayTeam"]

        gol_kandang = int(laga["FTHG"])
        gol_tandang = int(laga["FTAG"])

        if gol_kandang > gol_tandang:
            poin_kandang, poin_tandang = 3, 0
        elif gol_kandang < gol_tandang:
            poin_kandang, poin_tandang = 0, 3
        else:
            poin_kandang, poin_tandang = 1, 1

        riwayat_tim[kandang].append({
            "goals_for": gol_kandang,
            "goals_against": gol_tandang,
            "points": poin_kandang
        })

        riwayat_tim[tandang].append({
            "goals_for": gol_tandang,
            "goals_against": gol_kandang,
            "points": poin_tandang
        })

# Ubah hasil menjadi DataFrame
epl_features = pd.DataFrame(data_fitur)

if epl_features.empty:
    raise ValueError(
        "Belum ada baris fitur yang terbentuk. "
        "Periksa jumlah dan rentang data pertandingan."
    )

# Urutkan hasil fitur
epl_features = epl_features.sort_values("Date").reset_index(drop=True)

print("Pembuatan fitur selesai!")
print("Jumlah pertandingan awal:", len(epl_clean))
print("Jumlah pertandingan dengan fitur lengkap:", len(epl_features))
print("Jumlah kolom fitur dan target:", len(epl_features.columns))

display(epl_features.head())
```

```python

# TAHAP 6.1 — MENYIAPKAN DATA UNTUK TRAINING

import pandas as pd
import os

# Enam fitur yang digunakan oleh model
kolom_fitur = [
    "HomeGoalsAvg5",
    "HomeConcededAvg5",
    "HomePointsAvg5",
    "AwayGoalsAvg5",
    "AwayConcededAvg5",
    "AwayPointsAvg5"
]

kolom_target = "FTR"

# Muat file jika variabel belum tersedia
if "train_data" not in globals() or "test_data" not in globals():
    if not os.path.exists("epl_train.csv") or not os.path.exists("epl_test.csv"):
        raise FileNotFoundError(
            "File training/testing tidak ditemukan. "
            "Pastikan Tahap 5 sudah selesai."
        )

    train_data = pd.read_csv("epl_train.csv")
    test_data = pd.read_csv("epl_test.csv")

# Pastikan semua kolom tersedia
for nama_data, data in [
    ("Training", train_data),
    ("Testing", test_data)
]:
    kolom_wajib = kolom_fitur + [kolom_target]
    kolom_hilang = [
        kolom for kolom in kolom_wajib
        if kolom not in data.columns
    ]

    if kolom_hilang:
        raise ValueError(
            f"Kolom pada data {nama_data} tidak ditemukan: {kolom_hilang}"
        )

    if data[kolom_wajib].isna().any().any():
        raise ValueError(
            f"Data {nama_data} masih memiliki nilai kosong."
        )

# Pisahkan fitur (X) dan target (y)
X_train = train_data[kolom_fitur].copy()
y_train = train_data[kolom_target].copy()

X_test = test_data[kolom_fitur].copy()
y_test = test_data[kolom_target].copy()

print("Data berhasil disiapkan.")
print("Ukuran X_train:", X_train.shape)
print("Ukuran y_train:", y_train.shape)
print("Ukuran X_test:", X_test.shape)
print("Ukuran y_test:", y_test.shape)

print("\nKelas hasil pada training:")
print(y_train.value_counts())

print("\nKelas hasil pada testing:")
print(y_test.value_counts())
```

```python

# TAHAP 6.2 — MELATIH MODEL LOGISTIC REGRESSION

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# Membuat pipeline:
# 1. StandardScaler menstandarkan skala fitur.
# 2. LogisticRegression mempelajari pola hasil pertandingan.
model_epl = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(
        max_iter=2000,
        random_state=42
    ))
])

# Latih model HANYA menggunakan data training
model_epl.fit(X_train, y_train)

print("====================================")
print("TRAINING MODEL SELESAI")
print("====================================")
print("Algoritma: Logistic Regression")
print("Jumlah data training:", len(X_train))
print("Jumlah fitur:", len(kolom_fitur))
print("Kelas yang dipelajari:", model_epl.classes_)
```

```python

# TAHAP 6.3 — MEMBUAT PREDIKSI DAN MENGUKUR AKURASI

# MODEL PEMBANDING — MENANGANI KETIDAKSEIMBANGAN KELAS

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

model_epl_balanced = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(
        max_iter=2000,
        random_state=42,
        class_weight="balanced"
    ))
])

# Training hanya menggunakan data training
model_epl_balanced.fit(X_train, y_train)

# Evaluasi pada data testing yang sama
y_pred_balanced = model_epl_balanced.predict(X_test)

print("Akurasi:", round(accuracy_score(y_test, y_pred_balanced), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_balanced,
    labels=["H", "D", "A"],
    target_names=[
        "Menang Kandang",
        "Seri",
        "Menang Tandang"
    ],
    zero_division=0
))

print("\nConfusion Matrix:")
display(pd.DataFrame(
    confusion_matrix(
        y_test,
        y_pred_balanced,
        labels=["H", "D", "A"]
    ),
    index=["Aktual H", "Aktual D", "Aktual A"],
    columns=["Prediksi H", "Prediksi D", "Prediksi A"]
))
```

```python

# TAHAP 6.4 — MEMBANDINGKAN PREDIKSI DENGAN HASIL AKTUAL

# Gunakan hasil prediksi dari model seimbang
hasil_prediksi = test_data[
    ["Date", "HomeTeam", "AwayTeam"]
].copy()

hasil_prediksi["Hasil_Aktual"] = y_test.to_numpy()

# Perbedaan penting: gunakan y_pred_balanced
hasil_prediksi["Hasil_Prediksi"] = y_pred_balanced

# Terjemahkan kode hasil pertandingan
terjemahan_hasil = {
    "H": "Menang Kandang",
    "D": "Seri",
    "A": "Menang Tandang"
}

hasil_prediksi["Hasil_Aktual"] = (
    hasil_prediksi["Hasil_Aktual"].map(terjemahan_hasil)
)

hasil_prediksi["Hasil_Prediksi"] = (
    hasil_prediksi["Hasil_Prediksi"].map(terjemahan_hasil)
)

# Periksa apakah prediksi sesuai dengan hasil aktual
hasil_prediksi["Prediksi_Benar"] = (
    hasil_prediksi["Hasil_Aktual"]
    == hasil_prediksi["Hasil_Prediksi"]
)

# Tampilkan 15 pertandingan pertama
display(hasil_prediksi.head(15))

# Hitung hasil evaluasi
print(
    "Jumlah prediksi benar:",
    hasil_prediksi["Prediksi_Benar"].sum()
)

print(
    "Jumlah prediksi salah:",
    (~hasil_prediksi["Prediksi_Benar"]).sum()
)

print(
    "Akurasi:",
    round(hasil_prediksi["Prediksi_Benar"].mean() * 100, 2),
    "%"
)
```

```python

# TAHAP 7.2 — MEMBUAT FUNGSI PREDIKSI PERTANDINGAN

import pandas as pd

def prediksi_pertandingan(
    home_goals_avg5,
    home_conceded_avg5,
    home_points_avg5,
    away_goals_avg5,
    away_conceded_avg5,
    away_points_avg5
):
    """
    Memprediksi hasil pertandingan EPL berdasarkan
    rata-rata performa kedua tim dalam lima pertandingan
    sebelumnya.

    Hasil:
    H = Menang Kandang
    D = Seri
    A = Menang Tandang
    """

    # 1. Siapkan input sesuai urutan fitur saat pelatihan
    data_input = pd.DataFrame([{
        "HomeGoalsAvg5": home_goals_avg5,
        "HomeConcededAvg5": home_conceded_avg5,
        "HomePointsAvg5": home_points_avg5,
        "AwayGoalsAvg5": away_goals_avg5,
        "AwayConcededAvg5": away_conceded_avg5,
        "AwayPointsAvg5": away_points_avg5
    }], columns=kolom_fitur)

    # 2. Prediksi kelas hasil pertandingan
    hasil_kode = model_prediksi.predict(data_input)[0]

    # 3. Hitung probabilitas untuk setiap kelas
    probabilitas = model_prediksi.predict_proba(data_input)[0]

    # 4. Terjemahkan kode hasil menjadi bahasa Indonesia
    terjemahan_hasil = {
        "H": "Menang Kandang",
        "D": "Seri",
        "A": "Menang Tandang"
    }

    # 5. Susun probabilitas sesuai kelas model
    probabilitas_hasil = {
        terjemahan_hasil[kelas]: float(nilai)
        for kelas, nilai in zip(
            model_prediksi.classes_,
            probabilitas
        )
    }

    # 6. Kembalikan seluruh hasil prediksi
    return {
        "kode_hasil": hasil_kode,
        "hasil_prediksi": terjemahan_hasil[hasil_kode],
        "probabilitas": probabilitas_hasil
    }

print("Fungsi prediksi berhasil dibuat.")
```

```python

# TAHAP 8.2 — FUNGSI PENGHITUNGAN STATISTIK LIMA PERTANDINGAN

import pandas as pd

def hitung_statistik_lima_pertandingan(
    nama_tim,
    tanggal_prediksi,
    data=epl_historis
):
    """
    Menghitung statistik performa sebuah tim berdasarkan
    lima pertandingan terakhir sebelum tanggal prediksi.

    Statistik yang dihitung:
    - Rata-rata gol dicetak
    - Rata-rata gol kebobolan
    - Rata-rata poin

    Fungsi menggunakan pertandingan kandang maupun tandang.
    """

    # 1. Pastikan tanggal prediksi memiliki format datetime
    tanggal_prediksi = pd.to_datetime(tanggal_prediksi)

    # 2. Ambil pertandingan yang berlangsung SEBELUM
    #    tanggal prediksi, bukan pada tanggal yang sama
    data_sebelumnya = data[
        data["Date"] < tanggal_prediksi
    ].copy()

    # 3. Cari pertandingan ketika tim bermain sebagai tuan rumah
    pertandingan_kandang = data_sebelumnya[
        data_sebelumnya["HomeTeam"] == nama_tim
    ].copy()

    pertandingan_kandang["Gol_Dicetak"] = (
        pertandingan_kandang["FTHG"]
    )
    pertandingan_kandang["Gol_Kebobolan"] = (
        pertandingan_kandang["FTAG"]
    )

    # Poin tim kandang: menang 3, seri 1, kalah 0
    pertandingan_kandang["Poin"] = (
        pertandingan_kandang["FTR"]
        .map({"H": 3, "D": 1, "A": 0})
    )

    # 4. Cari pertandingan ketika tim bermain sebagai tim tamu
    pertandingan_tandang = data_sebelumnya[
        data_sebelumnya["AwayTeam"] == nama_tim
    ].copy()

    pertandingan_tandang["Gol_Dicetak"] = (
        pertandingan_tandang["FTAG"]
    )
    pertandingan_tandang["Gol_Kebobolan"] = (
        pertandingan_tandang["FTHG"]
    )

    # Poin tim tamu: menang 3, seri 1, kalah 0
    pertandingan_tandang["Poin"] = (
        pertandingan_tandang["FTR"]
        .map({"A": 3, "D": 1, "H": 0})
    )

    # 5. Gabungkan riwayat kandang dan tandang
    riwayat_tim = pd.concat(
        [pertandingan_kandang, pertandingan_tandang],
        ignore_index=True
    )

    # 6. Urutkan dari pertandingan terbaru ke terlama
    riwayat_tim = riwayat_tim.sort_values(
        "Date",
        ascending=False
    )

    # 7. Ambil tepat lima pertandingan terakhir
    lima_pertandingan = riwayat_tim.head(5).copy()

    # 8. Pastikan tersedia lima pertandingan yang valid
    if len(lima_pertandingan) < 5:
        raise ValueError(
            f"Tim '{nama_tim}' hanya memiliki "
            f"{len(lima_pertandingan)} pertandingan valid "
            f"sebelum {tanggal_prediksi.date()}. "
            "Diperlukan minimal 5 pertandingan."
        )

    kolom_statistik = [
        "Gol_Dicetak",
        "Gol_Kebobolan",
        "Poin"
    ]

    if lima_pertandingan[kolom_statistik].isnull().any().any():
        raise ValueError(
            f"Data statistik lima pertandingan terakhir "
            f"tim '{nama_tim}' mengandung nilai kosong."
        )

    # 9. Hitung rata-rata dari lima pertandingan tersebut
    statistik = {
        "GoalsAvg5": lima_pertandingan["Gol_Dicetak"].mean(),
        "ConcededAvg5": lima_pertandingan["Gol_Kebobolan"].mean(),
        "PointsAvg5": lima_pertandingan["Poin"].mean()
    }

    return statistik


print("Fungsi statistik lima pertandingan berhasil dibuat.")
```

```python

# TAHAP 8.2 — UJI COBA FUNGSI STATISTIK

# Ambil salah satu tim dari dataset
tim_uji = epl_historis["HomeTeam"].dropna().iloc[0]

# Gunakan tanggal setelah pertandingan terakhir dalam dataset
tanggal_uji = (
    epl_historis["Date"].max()
    + pd.Timedelta(days=1)
)

# Hitung statistik lima pertandingan sebelumnya
hasil_statistik = hitung_statistik_lima_pertandingan(
    nama_tim=tim_uji,
    tanggal_prediksi=tanggal_uji,
    data=epl_historis
)

print("Tim yang diuji:", tim_uji)
print("Tanggal prediksi uji:", tanggal_uji.date())

print("\nStatistik lima pertandingan sebelumnya:")
for nama_fitur, nilai in hasil_statistik.items():
    print(f"{nama_fitur}: {nilai:.2f}")
```

```python

# TAHAP 8.3 — MENGGABUNGKAN STATISTIK KEDUA TIM

def buat_fitur_pertandingan(
    tim_kandang,
    tim_tandang,
    tanggal_prediksi,
    data=epl_historis
):
    """
    Menghasilkan enam fitur input model dari statistik
    lima pertandingan sebelumnya untuk kedua tim.
    """

    # 1. Pastikan kedua tim tidak sama
    if tim_kandang == tim_tandang:
        raise ValueError(
            "Tim kandang dan tim tandang tidak boleh sama."
        )

    # 2. Hitung statistik lima pertandingan tim kandang
    statistik_kandang = hitung_statistik_lima_pertandingan(
        nama_tim=tim_kandang,
        tanggal_prediksi=tanggal_prediksi,
        data=data
    )

    # 3. Hitung statistik lima pertandingan tim tandang
    statistik_tandang = hitung_statistik_lima_pertandingan(
        nama_tim=tim_tandang,
        tanggal_prediksi=tanggal_prediksi,
        data=data
    )

    # 4. Gabungkan kedua statistik dengan nama fitur
    #    yang sama seperti saat model dilatih
    fitur = {
        "HomeGoalsAvg5": statistik_kandang["GoalsAvg5"],
        "HomeConcededAvg5": statistik_kandang["ConcededAvg5"],
        "HomePointsAvg5": statistik_kandang["PointsAvg5"],
        "AwayGoalsAvg5": statistik_tandang["GoalsAvg5"],
        "AwayConcededAvg5": statistik_tandang["ConcededAvg5"],
        "AwayPointsAvg5": statistik_tandang["PointsAvg5"]
    }

    # 5. Susun menjadi DataFrame satu baris
    #    dengan urutan fitur yang digunakan saat pelatihan
    data_fitur = pd.DataFrame(
        [fitur],
        columns=kolom_fitur
    )

    return data_fitur


print("Fungsi penggabungan enam fitur berhasil dibuat.")
```

```python

# TAHAP 8.4 — FUNGSI PREDIKSI OTOMATIS BERDASARKAN TIM DAN TANGGAL

def prediksi_otomatis(
    tim_kandang,
    tim_tandang,
    tanggal_prediksi,
    data=epl_historis
):
    """
    Memprediksi hasil pertandingan EPL berdasarkan:
    - Nama tim kandang
    - Nama tim tandang
    - Tanggal pertandingan yang ingin diprediksi

    Fungsi menghitung statistik lima pertandingan sebelumnya,
    membentuk enam fitur, lalu menjalankan model Logistic Regression.
    """

    # 1. Validasi nama tim
    if not isinstance(tim_kandang, str) or not tim_kandang.strip():
        raise ValueError("Nama tim kandang harus diisi.")

    if not isinstance(tim_tandang, str) or not tim_tandang.strip():
        raise ValueError("Nama tim tandang harus diisi.")

    if tim_kandang == tim_tandang:
        raise ValueError(
            "Tim kandang dan tim tandang tidak boleh sama."
        )

    # 2. Pastikan kedua tim tersedia dalam dataset
    daftar_tim = set(data["HomeTeam"].dropna()) | set(
        data["AwayTeam"].dropna()
    )

    if tim_kandang not in daftar_tim:
        raise ValueError(
            f"Tim kandang '{tim_kandang}' tidak ditemukan dalam dataset."
        )

    if tim_tandang not in daftar_tim:
        raise ValueError(
            f"Tim tandang '{tim_tandang}' tidak ditemukan dalam dataset."
        )

    # 3. Validasi tanggal prediksi
    tanggal_prediksi = pd.to_datetime(
        tanggal_prediksi,
        errors="coerce"
    )

    if pd.isna(tanggal_prediksi):
        raise ValueError(
            "Tanggal prediksi tidak valid. Gunakan format YYYY-MM-DD."
        )

    # 4. Buat enam fitur berdasarkan riwayat kedua tim
    fitur_pertandingan = buat_fitur_pertandingan(
        tim_kandang=tim_kandang,
        tim_tandang=tim_tandang,
        tanggal_prediksi=tanggal_prediksi,
        data=data
    )

    # 5. Jalankan prediksi kelas dan probabilitas
    kode_hasil = model_prediksi.predict(
        fitur_pertandingan
    )[0]

    probabilitas = model_prediksi.predict_proba(
        fitur_pertandingan
    )[0]

    # 6. Terjemahkan kode hasil pertandingan
    terjemahan_hasil = {
        "H": "Menang Kandang",
        "D": "Seri",
        "A": "Menang Tandang"
    }

    # 7. Hubungkan probabilitas dengan kelas model
    probabilitas_hasil = {
        terjemahan_hasil[kelas]: float(nilai)
        for kelas, nilai in zip(
            model_prediksi.classes_,
            probabilitas
        )
    }

    # 8. Kembalikan hasil lengkap
    return {
        "tim_kandang": tim_kandang,
        "tim_tandang": tim_tandang,
        "tanggal_prediksi": tanggal_prediksi.strftime("%Y-%m-%d"),
        "kode_hasil": kode_hasil,
        "hasil_prediksi": terjemahan_hasil[kode_hasil],
        "probabilitas": probabilitas_hasil,
        "fitur": fitur_pertandingan
    }


print("Fungsi prediksi otomatis berhasil dibuat.")
```
